```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/duochen/data-science-bootcamp/refs/heads/main/Data/gdp_development.csv")
print(df.head())
```

```python
print(df.describe())
```

```python
skew_df = pd.DataFrame(df.skew(numeric_only=True), columns=["Skewness"]).round(2)
print(skew_df)
```

format as a clean DataFrame


```python
sns.pairplot(df)
plt.suptitle("GDP and Development Indicators", y=1.02)
plt.show()
```

```python
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.title("Correlation Matrix")
plt.show()
```

```python
sns.regplot(
    x="GDP per Capita",
    y="Literacy Rate",
    data=df,
    scatter_kws={"s": 60, "alpha": 0.8}, # Adjusts point size and transparency
    line_kws={"color": "crimson"}       # Colors the trendline for visibility
)
plt.title("GDP per Capita vs. Literacy Rate", fontsize=14, fontweight="bold", pad=15)
plt.xlabel("GDP per Capita ($)", fontsize=12)
plt.ylabel("Literacy Rate (%)", fontsize=12)
```
